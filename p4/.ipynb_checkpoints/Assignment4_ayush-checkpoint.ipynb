```python
#import libraries
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt
import time # measures training time

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input
```

```python
#load the dataset
train = pd.read_csv("fashion-mnist_train.csv")
test = pd.read_csv("fashion-mnist_test.csv")

print("Train shape:", train.shape)
print("Test shape:", test.shape)
```

```python
#separate input and output
#x -> image pixels
#y -> class label
#axis=0 → operate down the rows → usually gives a result for each column
#axis=1 → operate across the columns → usually gives a result for each row

X_train = train.drop("label", axis=1)
y_train = train["label"]

X_test = test.drop("label", axis=1)
y_test = test["label"]
```

```python
#normalize the data
#converts pixel values from (0 to 255) to (o to 1)
X_train = X_train.astype("float32") / 255.0
X_test = X_test.astype("float32") / 255.0
```

```python
#Create the same MLP for all activations
def create_model(activation):

    model = Sequential([
        Input(shape=(784,)),                #Each input sample will contain 784 values

        Dense(256, activation=activation),  #first hidden layer
        Dense(128, activation=activation),  #second hidden layer
        Dense(64, activation=activation),   #third hidden layer

        Dense(10, activation="softmax")     #output layer with 10 neurons for 10 classes
    ])

    model.compile(
        optimizer="adam",
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model
```

```python
#Define activations and storage
activations = ["sigmoid", "tanh", "relu"]

#creates empty dictionaries
histories = {}     #stores accuracy and loss
times = {}         #stores training time
models = {}        #stores the training nn model
```

```python
#Train all three models
for activation in activations:

    print("\nTraining:", activation.upper())   #prints the activation function name

    model = create_model(activation)           #it creates a model using the current acitvation function 

    start = time.time()                        #records the current time before training starts.
                                               #here actual model training happens
    history = model.fit(                       #Train the nn using the training data
        X_train,                               #contains pixel values of image
        y_train,                               #contains the label of the image like shirt, jeans, etc
        epochs=10,                             #trains the complete dataset 10 times
        batch_size=128,                        #model doesn't process all training images at once. it processes 128 images before 1 weight update
        validation_split=0.2,                  #20% of the training data is used for validation during training only
        verbose=1                              #displays the training progress
    )

    end = time.time()                          #stores the time after training is finished

    histories[activation] = history            #stores the training history according to the activation function.
    times[activation] = end - start            #gives the total training time.
    models[activation] = model                 #stores the trained model itself.
```

```python
#Plot training speed
plt.figure(figsize=(5, 5))

plt.bar(
    [a.upper() for a in activations],
    [times[a] for a in activations]
)

plt.title("Training Speed Comparison")
plt.xlabel("Activation Function")
plt.ylabel("Training Time (seconds)")

# Zoom Y-axis
plt.ylim(min(times.values()) - 1, max(times.values()) + 1)

plt.show()
```

```python
#Plot training accuracy
plt.figure(figsize=(10, 6))

for activation in activations:

    plt.plot(
        histories[activation].history["accuracy"],
        label=activation.upper()
    )

plt.title("Training Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")

plt.legend()
plt.grid(True)

plt.show()
```

```python
#Plot validation accuracy
plt.figure(figsize=(10, 6))

for activation in activations:

    plt.plot(
        histories[activation].history["val_accuracy"],
        label=activation.upper()
    )

plt.title("Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")

plt.legend()
plt.grid(True)

plt.show()
```

```python
#Prepare sample data for gradient flow
#We take a small sample of 128 images to inspect gradients.
#We do not need the entire dataset just to measure gradient flow.
X_sample = tf.convert_to_tensor(
    X_train.iloc[:128].values,
    dtype=tf.float32
)

y_sample = tf.convert_to_tensor(
    y_train.iloc[:128].values
)
```

```python
#Calculate gradient flow
#A gradient tells us:
#How much should the weight change?

#We calculate the mean absolute gradient for each layer.
#Very small gradients can indicate a vanishing-gradient problem.
#Very large gradients can indicate an exploding-gradient problem.

gradient_data = {}

for activation in activations:

    model = models[activation]

    with tf.GradientTape() as tape:

        predictions = model(X_sample)

        loss = tf.keras.losses.sparse_categorical_crossentropy(
            y_sample,
            predictions
        )

        loss = tf.reduce_mean(loss)

    gradients = tape.gradient(
        loss,
        model.trainable_weights
    )

    values = []

    for gradient, weight in zip(
        gradients,
        model.trainable_weights
    ):

        if gradient is not None and "kernel" in weight.name:

            values.append(
                tf.reduce_mean(
                    tf.abs(gradient)
                ).numpy()
            )

    gradient_data[activation] = values
```

```python
#Plot gradient flow
layers = [
    "Layer 1",
    "Layer 2",
    "Layer 3",
    "Output"
]

plt.figure(figsize=(10, 6))

for activation in activations:

    plt.plot(
        layers,
        gradient_data[activation],
        marker="o",
        label=activation.upper()
    )

plt.title("Gradient Flow Comparison")
plt.xlabel("Layer")
plt.ylabel("Mean Absolute Gradient")

plt.legend()
plt.grid(True)

plt.show()
```
