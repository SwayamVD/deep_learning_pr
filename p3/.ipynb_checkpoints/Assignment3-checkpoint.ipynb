```python
import pandas as pd #reads dataset
import tensorflow as tf #creates and trains neural networks
import matplotlib.pyplot as plt # draw graphs

from tensorflow.keras.models import Sequential # creates the model layer by layer
from tensorflow.keras.layers import Dense, Dropout # dense - fully connected neural network, dropout - randomly disables neuron during training
from tensorflow.keras.regularizers import l2 # adds L2 regularization
```

```python
train = pd.read_csv("fashion-mnist_train.csv") # each image is 28x28 pixels
test = pd.read_csv("fashion-mnist_test.csv")

print("Train shape:", train.shape)
print("Test shape:", test.shape)

train.head()
```

```python
X_train = train.drop("label", axis=1) # contains image pixels
y_train = train["label"] # contains correct label

X_test = test.drop("label", axis=1)
y_test = test["label"]

print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)

print("X_test shape:", X_test.shape)
print("y_test shape:", y_test.shape)
```

```python
#normalize the pixel values from (o to 255) to (0 to 1)
X_train = X_train / 255.0
X_test = X_test / 255.0
```

```python
#Model 1 - without regularization 
# it has 256,128 and 64 neurons and 10 neurons at last as mnist has 10 image classes ans softmax gives probabilities for those 10 classes
model_normal = Sequential([
    
    Dense(
        256,
        activation="relu",
        input_shape=(784,)
    ),
    
    Dense(
        128,
        activation="relu"
    ),
    
    Dense(
        64,
        activation="tanh"
    ),
    
    Dense(
        10,
        activation="softmax"
    )
])
```

```python
# Compile the normal model
model_normal.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy", #suitable because you have multiple classes
    metrics=["accuracy"]
)

model_normal.summary()
```

```python
# Train the normal model
history_normal = model_normal.fit(
    X_train,
    y_train,
    epochs=10, #model sees the entire training dataset 10 times.
    batch_size=128, #model processes 128 images before updating its weights.
    validation_split=0.2 #80% → training, 20% → validation
)
```

```python
#Create the L2 model
model_l2 = Sequential([
    
    Dense(
        256,
        activation="relu",
        input_shape=(784,),
        kernel_regularizer=l2(0.001)
    ),
    
    Dense(
        128,
        activation="relu",
        kernel_regularizer=l2(0.001)
    ),
    
    Dense(
        64,
        activation="tanh",
        kernel_regularizer=l2(0.001)
    ),
    
    Dense(
        10,
        activation="softmax"
    )
])
```

```python
# compile and train the L2 model
model_l2.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model_l2.summary()

history_l2 = model_l2.fit(
    X_train,
    y_train,
    epochs=10,
    batch_size=128,
    validation_split=0.2
)
```

```python
# Dropout
model_dropout = Sequential([
    
    Dense(
        256,
        activation="relu",
        input_shape=(784,)
    ),
    
    Dropout(0.5),
    
    Dense(
        128,
        activation="relu"
    ),
    
    Dropout(0.5),
    
    Dense(
        64,
        activation="tanh"
    ),
    
    Dense(
        10,
        activation="softmax"
    )
])
```

```python
# compile and train the droupout model
model_dropout.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model_dropout.summary()

# TRAIN
history_dropout = model_dropout.fit(
    X_train,
    y_train,
    epochs=10,
    batch_size=128,
    validation_split=0.2
)
```

```python
# evaluate all three models
normal_loss, normal_accuracy = model_normal.evaluate(
    X_test,
    y_test,
    verbose=0
)

l2_loss, l2_accuracy = model_l2.evaluate(
    X_test,
    y_test,
    verbose=0
)

dropout_loss, dropout_accuracy = model_dropout.evaluate(
    X_test,
    y_test,
    verbose=0
)
```

```python
#print the results
print("Normal Model")
print("Test Loss:", normal_loss)
print("Test Accuracy:", normal_accuracy)

print("\nL2 Model")
print("Test Loss:", l2_loss)
print("Test Accuracy:", l2_accuracy)

print("\nDropout Model")
print("Test Loss:", dropout_loss)
print("Test Accuracy:", dropout_accuracy)
```

```python
#plot training and validation loss
plt.figure(figsize=(12, 6))

plt.plot(
    history_normal.history["loss"],
    label="Normal Training"
)

plt.plot(
    history_normal.history["val_loss"],
    linestyle="--",
    label="Normal Validation"
)

plt.plot(
    history_l2.history["loss"],
    label="L2 Training"
)

plt.plot(
    history_l2.history["val_loss"],
    linestyle="--",
    label="L2 Validation"
)

plt.plot(
    history_dropout.history["loss"],
    label="Dropout Training"
)

plt.plot(
    history_dropout.history["val_loss"],
    linestyle="--",
    label="Dropout Validation"
)

plt.title("Training and Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)

plt.show()
```

```python
#plot training and validation accuracy
plt.figure(figsize=(12, 6))

plt.plot(
    history_normal.history["accuracy"],
    label="Normal Training"
)

plt.plot(
    history_normal.history["val_accuracy"],
    linestyle="--",
    label="Normal Validation"
)

plt.plot(
    history_l2.history["accuracy"],
    label="L2 Training"
)

plt.plot(
    history_l2.history["val_accuracy"],
    linestyle="--",
    label="L2 Validation"
)

plt.plot(
    history_dropout.history["accuracy"],
    label="Dropout Training"
)

plt.plot(
    history_dropout.history["val_accuracy"],
    linestyle="--",
    label="Dropout Validation"
)

plt.title("Training and Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)

plt.show()
```

```python
#Compare final test accuracy
models = [
    "Normal",
    "L2",
    "Dropout"
]

accuracies = [
    normal_accuracy,
    l2_accuracy,
    dropout_accuracy
]

plt.figure(figsize=(8, 5))

bars = plt.bar(
    models,
    accuracies
)

plt.title("Test Accuracy Comparison")
plt.xlabel("Model")
plt.ylabel("Test Accuracy")

for bar, value in zip(bars, accuracies):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f"{value:.4f}",
        ha="center",
        va="bottom"
    )

plt.show()
```
