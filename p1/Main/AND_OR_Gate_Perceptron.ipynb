```python
import numpy as np
```

```python
# dataset

X = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1],
])

# AND GATE
# y = np.array([0,0,0,1])

# OR GATE
# y = np.array([0,1,1,1])

# NAND GATE
y = np.array([1,1,1,0])
```

```python
# hyperparameters

w = np.array([0.7,0.4])
b = 0.0
lr = 0.5
th = 1
epochs = 10
```

```python
for e in range(epochs):
    for i in range(len(X)):
        z = np.dot(X[i],w) + b
        if z >= th:
            pred = 1;
        else:
            pred = 0;

        err = y[i] - pred
        if err != 0:
            w = w + lr * err * X[i]
            b = b + lr * err 
```

```python
print("Final Updated Weights and Bias are: ", w, b)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(6,6))

plt.scatter(X[:, 0], X[:, 1], c=y)

x = np.linspace(-0.5, 1.5, 100)
y_line = (th - b - w[0] * x) / w[1]

plt.plot(x, y_line)
plt.xlabel("Input 1")
plt.ylabel("Input 2")
plt.title("AND Gate")
plt.grid()
plt.show()
```
