```python
# this is not real MLP its a simulation where we use 3 diff perceptrons

import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1],
])

and_y = np.array([0,0,0,1])
nand_y = np.array([1,1,1,0])
or_y = np.array([0,1,1,1])
```

```python
and_w = np.array([0.0,0.0])
nand_w = np.array([0.0,0.0])
or_w = np.array([0.0,0.0])

and_b = 0.0
nand_b = 0.0
or_b = 0.0

th = 1
lr = 0.5
epochs = 15
```

```python
def OR_gate(or_y, or_w, or_b):
    for e in range(epochs):
        for i in range(len(X)):
            or_z = np.dot(X[i],or_w) + or_b
            if or_z >= th:
                pred = 1
            else:
                pred = 0
    
            err = or_y[i] - pred
            if err != 0:
                or_w = or_w + lr * err * X[i]
                or_b = or_b + lr * err

    return or_w, or_b

def NAND_gate(nand_y, nand_w, nand_b):
    for e in range(epochs):
        for i in range(len(X)):
            nand_z = np.dot(X[i],nand_w) + nand_b
            if nand_z >= th:
                pred = 1
            else:
                pred = 0
    
            err = nand_y[i] - pred
            if err != 0:
                nand_w = nand_w + lr * err * X[i]
                nand_b = nand_b + lr * err
    return nand_w, nand_b

def AND_gate(and_y, and_w, and_b):
    for e in range(epochs):
        for i in range(len(X)):
            and_z = np.dot(X[i],and_w) + and_b
            if and_z >= th:
                pred = 1
            else:
                pred = 0
    
            err = and_y[i] - pred
            if err != 0:
                and_w = and_w + lr * err * X[i]
                and_b = and_b + lr * err
    return and_w, and_b
```

```python
or_w, or_b = OR_gate(or_y, or_w, or_b)
nand_w, nand_b = NAND_gate(nand_y, nand_w, nand_b)
and_w, and_b = AND_gate(and_y, and_w, and_b)
```

```python
def predict(X, w, b):
    z = np.dot(X, w) + b
    if z >= th:
        return 1
    else:
        return 0
    
```

```python
for i in range(len(X)):
    or_op = predict(X[i], or_w, or_b)
    nand_op = predict(X[i], nand_w, nand_b)
    new_inp = np.array([or_op,nand_op])
    xor_op = predict(new_inp, and_w, and_b)
    print(X[i], "->" , xor_op)
```
