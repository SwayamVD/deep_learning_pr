```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
import time

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Input,
    Flatten,
    Dense,
    Dropout,
    Conv2D,                     #CNN layer that detects image features such as edges, shapes, and patterns
    MaxPooling2D,               #Reduces feature-map size while keeping important information
    RandomRotation,             #Data augmentation: randomly rotates training images
    RandomZoom                  #Data augmentation: randomly zooms training images
)
```

```python
#Load mnist dataset
train = pd.read_csv("fashion-mnist_train.csv")
test = pd.read_csv("fashion-mnist_test.csv")

print("Train shape:", train.shape)
print("Test shape:", test.shape)

train.head()
```

```python
#Separate features and labels
X_train = train.drop("label", axis=1).values
y_train = train["label"].values

X_test = test.drop("label", axis=1).values
y_test = test["label"].values
```

```python
#Reshape images to 28 × 28
X_train = X_train.reshape(-1, 28, 28)
X_test = X_test.reshape(-1, 28, 28)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)

print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
#Normalize pixel values
X_train = X_train.astype("float32") / 255.0
X_test = X_test.astype("float32") / 255.0
```

```python
#Define class names
class_names = [
    "T-shirt/top",
    "Trouser",
    "Pullover",
    "Dress",
    "Coat",
    "Sandal",
    "Shirt",
    "Sneaker",
    "Bag",
    "Ankle boot"
]
```

```python
#Display sample images
plt.figure(figsize=(10, 5))

for i in range(10):

    plt.subplot(2, 5, i + 1)

    plt.imshow(
        X_train[i],
        cmap="gray"
    )

    plt.title(
        class_names[y_train[i]]
    )

    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
#Prepare CNN input shape
X_train_cnn = X_train.reshape(-1, 28, 28, 1)
X_test_cnn = X_test.reshape(-1, 28, 28, 1)

print("CNN train shape:", X_train_cnn.shape)
print("CNN test shape:", X_test_cnn.shape)
```

```python
#Define data augmentation - This slightly rotates and zooms training images to create more variety.
data_augmentation = Sequential([
    RandomRotation(0.1),
    RandomZoom(0.1)
])
```

```python
#Build the MLP model
mlp_model = Sequential([

    Input(shape=(28, 28)),

    Flatten(),

    Dense(
        256,
        activation="relu"
    ),

    Dense(
        128,
        activation="relu"
    ),

    Dropout(0.3),

    Dense(
        10,
        activation="softmax"
    )
])
```

```python
#Build the CNN model
cnn_model = Sequential([

    Input(shape=(28, 28, 1)),

    data_augmentation,

    Conv2D(
        32,
        kernel_size=(3, 3),
        activation="relu"
    ),

    MaxPooling2D(
        pool_size=(2, 2)
    ),

    Conv2D(
        64,
        kernel_size=(3, 3),
        activation="relu"
    ),

    MaxPooling2D(
        pool_size=(2, 2)
    ),

    Flatten(),

    Dense(
        128,
        activation="relu"
    ),

    Dropout(0.3),

    Dense(
        10,
        activation="softmax"
    )
])
```

```python
#Compile both models
mlp_model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

cnn_model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
#Check model summaries
print("MLP MODEL")
mlp_model.summary()

print("\nCNN MODEL")
cnn_model.summary()
```

```python
#Train the MLP
start = time.time()

history_mlp = mlp_model.fit(
    X_train,
    y_train,
    epochs=10,
    batch_size=128,
    validation_split=0.2,
    verbose=1
)

mlp_time = time.time() - start
```

```python
print(
    "MLP Training Time:",
    round(mlp_time, 2),
    "seconds"
)
```

```python
#Train the CNN
start = time.time()

history_cnn = cnn_model.fit(
    X_train_cnn,
    y_train,
    epochs=10,
    batch_size=128,
    validation_split=0.2,
    verbose=1
)

cnn_time = time.time() - start
```

```python
print(
    "CNN Training Time:",
    round(cnn_time, 2),
    "seconds"
)
```

```python
#Evaluate both models
mlp_loss, mlp_accuracy = mlp_model.evaluate(
    X_test,
    y_test,
    verbose=0
)

cnn_loss, cnn_accuracy = cnn_model.evaluate(
    X_test_cnn,
    y_test,
    verbose=0
)
```

```python
print("MLP Results")
print("Test Loss:", mlp_loss)
print("Test Accuracy:", mlp_accuracy)

print("\nCNN Results")
print("Test Loss:", cnn_loss)
print("Test Accuracy:", cnn_accuracy)
```

```python
#Compare accuracy, loss, and time
print("\nMODEL COMPARISON")
print("=" * 60)

print(
    f"MLP | Accuracy: {mlp_accuracy:.4f} | "
    f"Loss: {mlp_loss:.4f} | "
    f"Time: {mlp_time:.2f}s"
)

print(
    f"CNN | Accuracy: {cnn_accuracy:.4f} | "
    f"Loss: {cnn_loss:.4f} | "
    f"Time: {cnn_time:.2f}s"
)
```

```python
#Plot training and validation accuracy
plt.figure(figsize=(10, 6))

plt.plot(
    history_mlp.history["accuracy"],
    label="MLP Training"
)

plt.plot(
    history_mlp.history["val_accuracy"],
    linestyle="--",
    label="MLP Validation"
)

plt.plot(
    history_cnn.history["accuracy"],
    label="CNN Training"
)

plt.plot(
    history_cnn.history["val_accuracy"],
    linestyle="--",
    label="CNN Validation"
)

plt.title("MLP vs CNN Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")

plt.legend()
plt.grid(True)

plt.show()
```

```python
#Plot training and validation loss
plt.figure(figsize=(10, 6))

plt.plot(
    history_mlp.history["loss"],
    label="MLP Training"
)

plt.plot(
    history_mlp.history["val_loss"],
    linestyle="--",
    label="MLP Validation"
)

plt.plot(
    history_cnn.history["loss"],
    label="CNN Training"
)

plt.plot(
    history_cnn.history["val_loss"],
    linestyle="--",
    label="CNN Validation"
)

plt.title("MLP vs CNN Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")

plt.legend()
plt.grid(True)

plt.show()
```

```python
#Plot test accuracy comparison
models = [
    "MLP",
    "CNN"
]

accuracies = [
    mlp_accuracy,
    cnn_accuracy
]

plt.figure(figsize=(7, 5))

bars = plt.bar(
    models,
    accuracies
)

plt.title("Test Accuracy Comparison")
plt.ylabel("Accuracy")

for bar, value in zip(bars, accuracies):

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f"{value:.4f}",
        ha="center",
        va="bottom"
    )

plt.show()
```

```python
#Plot training time comparison
training_times = [
    mlp_time,
    cnn_time
]

plt.figure(figsize=(7, 5))

bars = plt.bar(
    models,
    training_times
)

plt.title("Training Time Comparison")
plt.ylabel("Time (seconds)")

for bar, value in zip(
    bars,
    training_times
):

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f"{value:.2f}s",
        ha="center",
        va="bottom"
    )

plt.show()
```

```python
#Generate CNN predictions
predictions = cnn_model.predict(
    X_test_cnn[:10]
)

predicted_classes = np.argmax(
    predictions,
    axis=1
)
```

```python
#Display actual and predicted classes
plt.figure(figsize=(12, 5))

for i in range(10):

    plt.subplot(
        2,
        5,
        i + 1
    )

    plt.imshow(
        X_test[i],
        cmap="gray"
    )

    actual = class_names[
        y_test[i]
    ]

    predicted = class_names[
        predicted_classes[i]
    ]

    plt.title(
        f"Actual: {actual}\nPred: {predicted}"
    )

    plt.axis("off")

plt.tight_layout()
plt.show()
```
